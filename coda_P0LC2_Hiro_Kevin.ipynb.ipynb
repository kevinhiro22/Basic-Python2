````python
#fungsi dibawah ini ditujukan untuk menhitung penghasilan menggunakan gaji per bulan berdasarkan jenis karyawan ditambah dengan jam lembur jika melebihi 40 jam per minggu
#rumus if digunakan untuk memilah jenis karyawan mana yang mendapatkan tambahan jam lembur
#fungsi return digunakan untuk mengunci jawaban agar dapat digunakan di soal no 2

def hitung_penghasilan(jenis_karyawan, jam_kerja):
    gaji_per_bulan = {'manager':10000000, 'staff':5000000, 'kasir':4000000}
    gaji = gaji_per_bulan[jenis_karyawan]

    if jam_kerja > 40:
         jam_lembur = (jam_kerja - 40) * 20000
    else:
        jam_lembur = 0

    total_pengeluaran = gaji + jam_lembur
    return total_pengeluaran

print(hitung_penghasilan('manager', 45))
print(hitung_penghasilan('staff', 42))
print(hitung_penghasilan('kasir', 38))
print(hitung_penghasilan('staff', 50))
````

````python
# fungsi dibawah ini ditujukan untuk daftar karyawan yang dibuat menggunakan dictionary
#fungsi def total pengeluaran digunakan untuk menghitung total gaji seluruh karyawan.
#fungsi for digunakan untuk looping jumlah karyawan dikarenakan daftar karyawan yang terbatas maka digunakan for
#fungsi return digunakan untuk melanjutkan hasil kalkulasi dari nomor 1

daftar_karyawan = [
             {'tingkat': 'manager', 'jam_kerja': 45},
             {'tingkat': 'staff', 'jam_kerja': 42},
             {'tingkat': 'kasir', 'jam_kerja': 38},
             {'tingkat': 'staff', 'jam_kerja': 50}
            ]

def total_pengeluaran(daftar_karyawan):
  total_pengeluaran = 0

  for karyawan in daftar_karyawan:
    jenis_karyawan = karyawan['tingkat']
    jam_kerja = karyawan['jam_kerja']
    total_pengeluaran += hitung_penghasilan(jenis_karyawan, jam_kerja)

  return total_pengeluaran

print(total_pengeluaran(daftar_karyawan))
````
